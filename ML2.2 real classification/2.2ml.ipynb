#### 2. Попробуйте применить к той же задаче другие модели классификации. Для каждой из них выведите матрицу классификации и оценку точности.

```python
from sklearn import metrics
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt
import seaborn as sns


def fit_report(model, X, y, title):
    model.fit(X, y)
    y_pred = model.predict(X)
    cm = metrics.confusion_matrix(y, y_pred)
    sns.heatmap(cm, annot=True, fmt="d", cmap="Blues")
    plt.title(title)
    plt.xlabel("Predicted")
    plt.ylabel("Actual")
    plt.show()
    acc = metrics.accuracy_score(y, y_pred)
    print(f"Accuracy: {acc:.4f}")
```

#### Метод опорных векторов:

#### a. Без ядра

```python
from sklearn.svm import LinearSVC

svm_no_kernel = make_pipeline(StandardScaler(), LinearSVC(max_iter=5000))
fit_report(svm_no_kernel, X, y, "LinearSVC (no kernel)")
```

#### b. С линейным ядром

```python
from sklearn.svm import SVC

svm_linear = make_pipeline(StandardScaler(), SVC(kernel="linear"))
fit_report(svm_linear, X, y, "Linear kernel")
```

#### c. С гауссовым ядром

```python
from sklearn.svm import SVC

svm_rbf = make_pipeline(StandardScaler(), SVC(kernel="rbf", gamma="scale"))
fit_report(svm_rbf, X, y, "RBF kernel")
```

#### d. С полиномиальным ядром

```python
from sklearn.svm import SVC

svm_poly = make_pipeline(StandardScaler(), SVC(kernel="poly", degree=3, gamma="scale"))
fit_report(svm_poly, X, y, "Polynomial kernel")
```

#### Метод ближайших соседей

```python
from sklearn.neighbors import KNeighborsClassifier

knn = make_pipeline(StandardScaler(), KNeighborsClassifier(n_neighbors=5))
fit_report(knn, X, y, "KNN")
```

#### Многослойный перцептрон

```python
from sklearn.neural_network import MLPClassifier

mlp = make_pipeline(
    StandardScaler(),
    MLPClassifier(hidden_layer_sizes=(50, 50), max_iter=2000, random_state=42)
)
fit_report(mlp, X, y, "MLP")
```

#### Дерево решений

```python
from sklearn.tree import DecisionTreeClassifier

tree = DecisionTreeClassifier(random_state=42)
fit_report(tree, X, y, "Decision tree")
```

#### Наивный байесовский классификатор

```python
from sklearn.naive_bayes import GaussianNB

nb = GaussianNB()
fit_report(nb, X, y, "Gaussian naive bayes")
```

#### Другие методы:

#### a. Пассивно-агрессивный классификатор

```python
from sklearn.linear_model import PassiveAggressiveClassifier

pa = make_pipeline(StandardScaler(), PassiveAggressiveClassifier(max_iter=2000, random_state=42))
fit_report(pa, X, y, "Passive-aggressive")
```

#### b. Гребневый классификатор

```python
from sklearn.linear_model import RidgeClassifier

ridge = make_pipeline(StandardScaler(), RidgeClassifier())
fit_report(ridge, X, y, "Ridge classifier")
```

#### c. Случайный лес

```python
from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier(n_estimators=200, random_state=42)
fit_report(rf, X, y, "Random forest")
```

#### d. Беггинг

```python
from sklearn.ensemble import BaggingClassifier

bag = BaggingClassifier(n_estimators=200, random_state=42)
fit_report(bag, X, y, "Bagging")
```

#### e. Другие модели по желанию (Gradient Boosting)

```python
from sklearn.ensemble import GradientBoostingClassifier

gb = GradientBoostingClassifier(random_state=42)
fit_report(gb, X, y, "Gradient boosting")
```

#### 3. Напишите функцию, которая автоматически обучает все перечисленные модели и для каждой выдает оценку точности.

```python
def train_all_models(X, y):
    models = {
        "LinearSVC": make_pipeline(StandardScaler(), LinearSVC(max_iter=5000)),
        "SVC linear": make_pipeline(StandardScaler(), SVC(kernel="linear")),
        "SVC rbf": make_pipeline(StandardScaler(), SVC(kernel="rbf", gamma="scale")),
        "SVC poly": make_pipeline(
            StandardScaler(), SVC(kernel="poly", degree=3, gamma="scale")
        ),
        "KNN": make_pipeline(StandardScaler(), KNeighborsClassifier(n_neighbors=5)),
        "MLP": make_pipeline(
            StandardScaler(),
            MLPClassifier(hidden_layer_sizes=(50, 50), max_iter=2000, random_state=42),
        ),
        "Decision Tree": DecisionTreeClassifier(random_state=42),
        "GaussianNB": GaussianNB(),
        "Passive Aggressive": make_pipeline(
            StandardScaler(), PassiveAggressiveClassifier(max_iter=2000, random_state=42)
        ),
        "Ridge": make_pipeline(StandardScaler(), RidgeClassifier()),
        "Random Forest": RandomForestClassifier(n_estimators=200, random_state=42),
        "Bagging": BaggingClassifier(n_estimators=200, random_state=42),
        "Gradient Boosting": GradientBoostingClassifier(random_state=42),
    }

    results = []
    for name, model in models.items():
        model.fit(X, y)
        y_pred = model.predict(X)
        acc = metrics.accuracy_score(y, y_pred)
        results.append((name, acc))

    results_df = pd.DataFrame(results, columns=["model", "accuracy"]).sort_values(
        "accuracy", ascending=False
    )
    return results_df

train_all_models(X, y)
```
