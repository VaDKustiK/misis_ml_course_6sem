#### 3. При выполнении третьего задания убедитесь более явно, что обильное количество пропущенных значений не является ошибкой объединения. Проведите больше, чем одну точечную проверку.

```python
if 'data_with_orders' not in globals():
    if 'data_123' not in globals():
        raise RuntimeError('data_123 not created yet')
    if 'orders_data' not in globals():
        orders_data = pd.read_csv('data/orders.csv', index_col=0)
        orders_data = orders_data.rename(columns={'Id': 'Order_id'})
    data_with_orders = data_123.merge(orders_data, on='Order_id', how='left')

order_ids_in_orders = set(orders_data['Order_id'].dropna())

has_order = data_with_orders['Order_id'].notna()
has_order_info = data_with_orders['order_date_time'].notna()

matched = (has_order & has_order_info).sum()
unmatched_with_order = (has_order & ~has_order_info).sum()
no_order_id = (~has_order).sum()

print(f'Есть Order_id и найдена запись в orders: {matched}')
print(f'Есть Order_id, но записи в orders нет: {unmatched_with_order}')
print(f'Order_id отсутствует изначально: {no_order_id}')

sample_missing = data_with_orders.loc[has_order & ~has_order_info, 'Order_id'].dropna().head(5).tolist()
sample_found = data_with_orders.loc[has_order & has_order_info, 'Order_id'].dropna().head(5).tolist()

print('\nТочечные проверки (несколько примеров):')
print('Примеры, где данных заказа нет (ожидаем False):')
for oid in sample_missing:
    print(f'  {oid} -> в orders_data: {oid in order_ids_in_orders}')

print('Примеры, где данные заказа есть (ожидаем True):')
for oid in sample_found:
    print(f'  {oid} -> в orders_data: {oid in order_ids_in_orders}')
```

#### 4. При выполнении третьего задания попробуйте использовать разные виды соединений. Как это отражается на структуре датасета?

```python
import pandas as pd

if 'data_123' not in globals():
    raise RuntimeError('Сначала выполните задание 1 (там готовится data_123).')

if 'orders_data' not in globals():
    orders_data = pd.read_csv('data/orders.csv', index_col=0)
    orders_data = orders_data.rename(columns={'Id': 'Order_id'})

join_types = ['left', 'inner', 'right', 'outer']
summary = []

for how in join_types:
    merged = data_123.merge(orders_data, on='Order_id', how=how)
    summary.append({
        'join_type': how,
        'rows': len(merged),
        'cols': merged.shape[1],
        'rows_with_order_info': merged['order_date_time'].notna().sum(),
        'missing_order_info': merged['order_date_time'].isna().sum(),
    })

join_summary = pd.DataFrame(summary)
print('Сравнение типов соединения:')
join_summary
```

```python
# Для основной задачи интеграции обращений с заказами оптимален left join: он сохраняет все обращения из data_123
# inner join сильно уменьшает выборку (остаются только записи с найденным заказом), поэтому часть обращений теряется
# right и outer добавляют строки из таблицы заказов, которых нет в обращениях, и меняют структуру выборки
# Поэтому для дальнейшего анализа качества поддержки логично использовать left join как базовый вариант
```
